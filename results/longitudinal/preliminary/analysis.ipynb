```python
import os
import pandas as pd
import sys

print(pd.__version__)
```

```python
results_file = 'preliminary_results.csv'

if not os.path.exists(results_file):
    print(f'File {results_file} not found.')
    sys.exit(-1)

df = pd.read_csv(results_file, sep='|')
```

```python
df.mean(numeric_only=True).round(0).to_frame()
```

```python
df.median(numeric_only=True).round(0).to_frame()
```

```python
total_commits = 6839
```

```python
df.mean(numeric_only=True).to_frame().multiply(total_commits).divide(60 * 1000).round(0)
```
